```python
batsman= pd.read_csv('sharma-kohli.csv')
batsman
```

```python
plt.plot(batsman['index'], batsman['RG Sharma'])
plt.plot(batsman['index'], batsman['V Kohli'])
```

```python
plt.plot(batsman['index'], batsman['RG Sharma'])
plt.plot(batsman['index'], batsman['V Kohli'])

plt.title('V kohli and RG sharma career comparison')
plt.xlabel('Season')
plt.ylabel('Run Scored')
```

```python
plt.plot(batsman['index'], batsman['RG Sharma'], color='red', linestyle='dashed', linewidth=3)
plt.plot(batsman['index'], batsman['V Kohli'], color='green', linestyle='solid', linewidth=1)

plt.title('V kohli and RG sharma career comparison')
plt.xlabel('Season')
plt.ylabel('Run Scored')
```

```python
plt.plot(batsman['index'], batsman['RG Sharma'], color='red', linestyle='dashed', linewidth=3, marker='D',  markersize=7)
plt.plot(batsman['index'], batsman['V Kohli'], color='green', linestyle='solid', linewidth=1, marker='>')

plt.title('V kohli and RG sharma career comparison')
plt.xlabel('Season')
plt.ylabel('Run Scored')
```

```python
plt.plot(batsman['index'], batsman['RG Sharma'], color='red', linestyle='dashed', linewidth=3, marker='D',  markersize=7, label='Rohit')
plt.plot(batsman['index'], batsman['V Kohli'], color='green', linestyle='solid', linewidth=1, marker='>', label='Kolhi')

plt.title('V kohli and RG sharma career comparison')
plt.xlabel('Season')
plt.ylabel('Run Scored')

plt.legend(loc='upper right')  #by default  plt.legend(loc='best')
```

```python
plt.plot(batsman['index'], batsman['RG Sharma'], color='red', linestyle='dashed', linewidth=3, marker='D',  markersize=7, label='Rohit')
plt.plot(batsman['index'], batsman['V Kohli'], color='green', linestyle='solid', linewidth=1, marker='>', label='Kolhi')

plt.title('V kohli and RG sharma career comparison')
plt.xlabel('Season')
plt.ylabel('Run Scored')

plt.legend()
plt.grid()
plt.show()
```

# Bar chart

```python
# simple bar chart
children = [10,20,40,10,30]
colors = ['red','blue','green','yellow','pink']

plt.bar(colors, children, color='red')
```

```python
plt.barh(colors, children, color='red')
```

```python
df=pd.read_csv("batsman_season_record.csv")
df
```

```python
plt.bar(df['batsman'], df['2015'])
```

```python
plt.bar(np.arange(df.shape[0])- 0.2, df['2015'], width=0.2, label='2015')
plt.bar(np.arange(df.shape[0]), df['2016'], width=0.2, label='2016')
plt.bar(np.arange(df.shape[0]) + 0.2, df['2017'], width=0.2, label='2017')

plt.xticks(np.arange(df.shape[0]), df['batsman']) #to replace with 0---4 to batsman name
plt.legend()
```

```python
# a problem
children = [10,20,40,10,30]
colors = ['red red red red red red','blue blue blue blue','green green green green green','yellow yellow yellow yellow ','pink pinkpinkpink']

plt.bar(colors,children,color='black')
plt.xticks(rotation='vertical')
plt.show()
```

```python
# Stacked Bar chart
plt.bar(df['batsman'], df['2015'], color='red', label='2015')
plt.bar(df['batsman'], df['2016'], bottom=df['2015'], color='green', label='2016')
plt.bar(df['batsman'], df['2017'], bottom=(df['2015']+df['2016']), color='purple', label='2017')

plt.legend()
```

# Pie Chart

```python
data = [23,45,100,20,49]
subjects = ['eng','science','maths','sst','hindi']
plt.pie(data, labels= subjects)
plt.show()
```

```python
df = pd.read_csv('gayle-175.csv')
df
```

```python
plt.pie(df['batsman_runs'], labels=df['batsman'], autopct='%0.1f%%')
plt.show()
```

```python
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',colors=['blue','green','yellow','pink','cyan','brown'])
plt.show()
```

```python
plt.pie(df['batsman_runs'], labels=df['batsman'], autopct='%0.1f%%', explode=[0.1,0,0,0,0.2,0], shadow=True)
```
