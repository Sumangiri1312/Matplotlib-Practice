```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
iris=pd.read_csv('iris.csv')
iris.sample(5)
```

```python
iris['Species']=iris['Species'].replace({'Iris-virginica':0, 'Iris-versicolor':1, 'Iris-setosa':2})
iris.sample(5)
```

```python
plt.scatter(iris['SepalLengthCm'], iris['PetalLengthCm'], c=iris['Species'], cmap='jet', alpha=0.6) #c--color and cmap=colormap
plt.xlabel('Sepla length')
plt.ylabel('Petal length')
plt.colorbar()
```

# Plot size

```python
plt.figure(figsize=(15,7))

plt.scatter(iris['SepalLengthCm'], iris['PetalLengthCm'], c=iris['Species'], cmap='jet', alpha=0.6)
plt.xlabel('Sepla length')
plt.ylabel('Petal length')
plt.colorbar()
```

# Annotations

```python
x = [1,2,3,4]
y = [5,6,7,8]

plt.scatter(x,y)
plt.text(1,5,'Point 1')
plt.text(2,6,'Point 2')
plt.text(3,7,'Point 3')
plt.text(4,8,'Point 4',fontdict={'size':12,'color':'brown'})
```

```python
df=pd.read_csv('batter.csv')
sample_df=df.head(100).sample(25, random_state=5)  #random_state use to pick same random number always
sample_df
```

```python
sample_df['batter'].values
```

```python
plt.figure(figsize=(15,9))
plt.scatter(sample_df['avg'], sample_df['strike_rate'], s=sample_df['runs']) #s=size
plt.xlabel('Average')
plt.ylabel('Strike rate')

plt.axhline(130, color='red')
plt.axvline(32, color='red')

for i in range(sample_df.shape[0]):
    if(i!=24):
        plt.text(sample_df['avg'].values[i], sample_df['strike_rate'].values[i], sample_df['batter'].values[i])
    else:
        plt.text(sample_df['avg'].values[i], sample_df['strike_rate'].values[i], sample_df['batter'].values[i], fontdict={'size':12,'color':'brown'})
```
