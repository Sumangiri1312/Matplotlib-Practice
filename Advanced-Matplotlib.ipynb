```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
iris=pd.read_csv('iris.csv')
iris.sample(5)
```

```python
iris['Species']=iris['Species'].replace({'Iris-virginica':0, 'Iris-versicolor':1, 'Iris-setosa':2})
iris.sample(5)
```

```python
plt.scatter(iris['SepalLengthCm'], iris['PetalLengthCm'], c=iris['Species'], cmap='jet', alpha=0.6) #c--color and cmap=colormap
plt.xlabel('Sepla length')
plt.ylabel('Petal length')
plt.colorbar()
```

```python
df=pd.read_csv('batter.csv')
sample_df=df.head(100).sample(25, random_state=5)  #random_state use to pick same random number always
sample_df
```

```python
sample_df['batter'].values
```

```python
plt.figure(figsize=(15,9))
plt.scatter(sample_df['avg'], sample_df['strike_rate'], s=sample_df['runs']) #s=size
plt.xlabel('Average')
plt.ylabel('Strike rate')

plt.axhline(130, color='red')
plt.axvline(32, color='red')

for i in range(sample_df.shape[0]):
    if(i!=24):
        plt.text(sample_df['avg'].values[i], sample_df['strike_rate'].values[i], sample_df['batter'].values[i])
    else:
        plt.text(sample_df['avg'].values[i], sample_df['strike_rate'].values[i], sample_df['batter'].values[i], fontdict={'size':12,'color':'brown'})
```

# 3D Surface Plots

```python
x=np.linspace(-10,10,100)
y=np.linspace(-10,10,100)

xx, yy=np.meshgrid(x,y)
```

```python
xx.shape
z=xx**2 + yy**2
z
```

```python
fig= plt.figure()

ax= plt.subplot(projection='3d')

ax.plot_surface(xx,yy,z, cmap='summer')
```

```python
fig= plt.figure()

ax= plt.subplot(projection='3d')

p=ax.plot_surface(xx,yy,z, cmap='summer')
fig.colorbar(p)
```

# Contour Plots = represent 3d surface into 2d

```python
fig= plt.figure()

ax= plt.subplot()

ax.contour(xx,yy,z, cmap='summer')
```

```python
fig= plt.figure()

ax= plt.subplot()

ax.contourf(xx,yy,z)
```

# Heatmap --(imshow())

```python
# find the most no of sixes that hitted in which over
delivery= pd.read_csv('IPL_Ball_by_Ball_2008_2022.csv')
delivery.head()
```

```python
temp_df=delivery[delivery['ballnumber'].isin([1,2,3,4,5,6]) & (delivery['batsman_run']==6)]
df=temp_df.pivot_table(index='overs', columns='ballnumber', values='batsman_run', aggfunc='count')
df
```

```python
plt.figure(figsize=(10,9))

plt.imshow(df) 
plt.colorbar()
```
